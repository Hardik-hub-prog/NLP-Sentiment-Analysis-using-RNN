```python
def remove_urls(text):
     text =  re.sub(r"http\S+" , "" , text) #(pattern , replacement , String) eg - https://www.google.com
     return text
    
df["review"] = df["review"].apply(remove_urls)
```

### Removing Puncutations

```python
def remove_punctuations(text):
     text = re.sub(r"[^A-Za-z0-9\s]" , "" , text) # exculde [A-Z , a-z , 0-9 , white spaces \s] apart from this punctuations(include) - replace empty space
     return text

df["review"] = df["review"].apply(remove_punctuations)
```

### Removing HTML

```python
def remove_html(text):
    text = re.sub(r"<.*?>" , "" , text) # ? - non-greedy manner remove , first closing tag -stop , remove HTML - space
    return text

df["review"] = df["review"].apply(remove_html)
```

```python
def remove_stopwd(text):
    tokens = word_tokenize(text)
    stop_word = stopwords.words("english")

    for word in tokens:
        if word in stop_word:
            text = text.replace(word , "")

    return text

df["review"] = df["review"].apply(remove_stopwd)
```

### Stemming

```python
from nltk.stem import PorterStemmer
```

```python
def stemming(text):
    ps = PorterStemmer()
    stem_word = []

    tokens = word_tokenize(text) # First tokenization perform
    for token in tokens:
        stem_token = ps.stem(token)  # convert each token into its root form
        stem_word.append(stem_token)  # empty list append new tokens

    return " ".join(stem_word)    # No return text(as return list , convert into text by " ".join)

df["review"] = df["review"].apply(stemming)
```

```python
 # Tensor Dataset - To define Dataset, access from Memory(RAM)
train_set = TensorDataset(

    torch.from_numpy(X_train).float(),  # Directly numpy array to tensor convert
    torch.from_numpy(y_train.values).float()
)

test_set = TensorDataset(

    torch.from_numpy(X_test).float(),
    torch.from_numpy(y_test.values).float()
)
```

```python
# Data Loader - To define how the data will be loaded for traning

train_loader = DataLoader(train_set , shuffle = True , batch_size = 64 ) 
test_loader = DataLoader(test_set , shuffle = True , batch_size = 64)
```

### Build our RNN

```python
import torch.nn as nn
import torch.optim as optim
```

```python
# Many to one RNN Architecture(Sentiment Analysis)
class RNN(nn.Module):
    def __init__(self , input_size , hidden_size = 128 , num_layer =1):
        super().__init__()

        self.hidden_size = hidden_size  # Attribute create for class RNN
        self.num_layer = num_layer

        #RNN layer (Pytorch RNN already define)
        self.rnn = nn.RNN(input_size , hidden_size , num_layer , batch_first = True)

        # fully connected layer
        self.fc = nn.Linear(hidden_size ,1) # single output

    def forward(self , x):
        # optional - shape(num_layer , batch_size , hidden_size)
        h0 = torch.zeros(self.num_layer , x.size(0), self.hidden_size)

        out , _ = self.rnn(x , h0)
        # 1st value = hidden state of all the timesteps => (batch, seq_len, hidden size)
        # 2nd value = final hidden state of last timestep

        out = self.fc(out[:, -1, :])
        return out

    
        

        

        
```

```python
input_size = X_train.shape[1]

model = RNN(input_size)  # object create , constructor call

criterion = nn.BCELoss() # Binary cross entropy loss for Binary classification
optimizer = optim.Adam(model.parameters())
```

### Traning and Evaluation

```python
epochs = 10

for epoch in range(epochs):
    model.train()
    running_loss = 0.0

    for Xb , yb in train_loader:

        optimizer.zero_grad()  # for each batch fresh gradient compute, dont accumualte value of graident of previous batch

        Xb = Xb.unsqueeze(1)  # add singleton direction - 2D - 3D as model expect 3D value - for that we used unsqueeze function

        output = model(Xb)  # output dimension - (batch_size ,1) 2D  Forward propagation
        # but here output is not a probaility value
        output = torch.sigmoid(output.squeeze())  # (batch_size, ) sequeeze dimension , convert o/p value - probaility

        loss = criterion(output , yb)  # Loss compute
        loss.backward() # Bckward propagation - gradient compute
        optimizer.step() # weights update

        running_loss += loss # Total running loss cal for all batches(620) in 1 epoch
        
    train_loss = running_loss/len(train_loader)  # Avg loss per batch for 1 epoch
        
    print(f"epoch = {epoch}/{epochs} , loss = {train_loss}")    
          
    
```

# Evaluate Model

```python
model.eval()

total =0
correct =0

with torch.no_grad():  # No gradiet compute(No BP) as we are doing testing not traning
    for Xb , yb in test_loader:
        

        Xb = Xb.unsqueeze(1)  # add singleton direction - 2D - 3D as model expect 3D value - for that we used unsqueeze function
        output = model(Xb) # FP but here again it is not probalility  - (batch_size , 1)

        predicted = (torch.sigmoid(output.squeeze()) > 0.5).float() # (batch_size, ) sequeeze dimension , convert o/p value - probaility

        total += yb.size(0)  # Total y_test - 9917

        correct += (predicted == yb).sum().item()  # sum store for all batches
```
